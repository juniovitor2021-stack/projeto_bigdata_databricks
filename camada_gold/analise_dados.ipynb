```python
from pyspark.sql.functions import *

#=======================================
# Objetivo dos análises
##=======================================

#1- Estados com maior número de acidentes 
#2 - Meses com maior número de ocorrencias de acidentes
#3 - Tipos de acidentes mais frequentes
```

```python
#=======================================
# Carregar tabela silver
#=======================================

df_gold = spark.table("projeto_profissional_bigdata.datatran_2018.silver_datatran18")
```

```python
#=======================================
# Visualizar dados
#=======================================
df_gold.show(5)
display(df_gold)
```

```python
#=======================================
# Pergunta 1 
# estados com maior número de acidenes
#=======================================

gold_acidente_por estado = (
    df_gold.("uf")
    .agg(count("*").alias("total_acidentes"))
    .orderBy(col("total_acidentes").desc())
)

print("ESTADOS COM MAIS ACIDENTES")
gold_acidente_por_estado.show()
```

```python
#=======================================
# Salvar tabela gold
#=======================================

gold_acidente_por_estado.write\
    .format("delta")
    .mode("overwrite")\
    .saveTable("projeto_profissional_bigdata")
```

```python
gold_acidente_por_mes = (
    df_gold.groupBy("mes_acidente", "nome_mes")
    .agg(count("*").alias("total_acidentes"))
    .orderBy(col("total_acidentes").
)

print("ESTADOS COM MAIS ACIDENTES")
gold_acidente_por_mes.show()
```

```python
#=======================================
# Salvar tabela gold mês
#=======================================

gold_acidente_por_estado.write\
    .format("delta")\
    .mode("overwrite")\
    .saveTable("projeto_profissional_bigdata_datatran_2018".gold_acidente_por_mes")
```

```python
#=======================================
# pergunta 3
# tipos de acidentes mais frequentes
#=======================================

gold_tipo_acidentes = (
    df_gold.groupBy("tipo_acidente")
    .agg(count("*").alias("total_acidentes"))
    .orderBy(col("total_acidentes").desc())
)

print("TIPOS DE ACIDENTES MAIS FREQUENTES")
display(gold_tipos_acidentes)
```

```python
gold_acidente_por_estado.write\
    .format("delta")\
    .mode("overwrite")\
    .saveTable("projeto_profissional_bigdata_datatran_2018".gold_tipo_acidentes")
```
